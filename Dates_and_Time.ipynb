```python
import pandas as pd

dates = pd.DataFrame({
    "Date": ["2026-01-05", "2026-02-10", "2026-03-15", "2026-04-20"]
})

dates
```

```python
#CONVERTING THE COLUMN TO DATE
dates["Date"] = pd.to_datetime(dates["Date"])
```

```python
#checking the data type
dates.dtypes
```

```python
#extract the year
dates["Year"] = dates["Date"].dt.year
```

```python
dates
```

```python
#extract the month
dates["Month"] = dates["Date"].dt.month
```

```python
#extract the day
dates["Day"] = dates["Date"].dt.day
```

```python
#extract the day of the week
dates["Day_Name"] = dates["Date"].dt.day_name()
```

```python
#extract the month name
dates["Month_Name"] = dates["Date"].dt.month_name()
```

```python
dates
```

```python
#creating a more realistic dataset
sales = pd.DataFrame({
    "Date": [
        "2026-01-05",
        "2026-01-15",
        "2026-02-10",
        "2026-02-20",
        "2026-03-05",
        "2026-03-18"
    ],

    "Sales": [100, 150, 200, 180, 250, 300]
})

sales
```

```python
#converting the date column
sales["Date"] = pd.to_datetime(sales["Date"])
```

```python
#add the month
sales["Month"] = sales["Date"].dt.month_name()
```

```python
sales
```

```python
#group by month
#this gives us total sales for each month
sales.groupby("Month")["Sales"].sum()
```

```python
#sort the results
#to find the highest_sales month
sales.groupby("Month")["Sales"].sum().sort_values(ascending=False)
```

```python
#group by day of the week
sales["Day"] = sales["Date"].dt.day_name()
```

```python
sales.groupby("Day")["Sales"].sum().sort_values(ascending=False)
```

```python
#finging the difference between dates
start = pd.to_datetime("2026-01-01")
end = pd.to_datetime("2026-01-15")
```

```python
end - start
```

```python
#filtering by dates
#suppose we want  sales that occured after January 31
sales[sales["Date"]> "2026-01-31"]
```

```python
#sales during January
sales[sales["Date"].dt.month == 1]
```

```python
import pandas as pd
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
df.head()
```

```python
df['Age'].corr(df['Fare'])
```

Age - Fare = 0.1

```python
df['Age'].corr(df['SibSp'])
```

Age -Sibsp = -0.3

```python
df['Fare'].corr(df['Pclass'])
```

Fare - Sibsp = -0.5

```python
df[['Age', 'Fare', 'SibSp', 'Parch']].corr()
```

```python
import seaborn as sns
sns.heatmap(df[['Age', 'Fare', 'SibSp', 'Parch']].corr(), annot=True)
```

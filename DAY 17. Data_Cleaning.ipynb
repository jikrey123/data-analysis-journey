```python
import pandas as pd
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
df.head()
```

```python
#checking for the data structure
df.info()
```

```python
#missing values for each cell
df.isnull()
```

```python
#missing values for a large dataset
df.isnull().sum()
```

```python
#finding the percentage of missing values
df.isnull().sum() * 100
```

```python
#to remove passengers whose age is missing
df.dropna(subset=["Age"])
```

```python
#filling age with the median
df["Age"]= df["Age"].fillna(df["Age"].median())
```

```python
#to check whether age is still missing
df["Age"].isnull().sum()
```

```python
#finding duplicate rows
df.duplicated().sum()
```

```python
#to remove duplicate rows
df = df.drop_duplicates()
```

```python
#checking again for duplicates
df.duplicated().sum()
```

```python
#checking data types
df.dtypes
```

```python
#HANDLING CATEGORICAL MISSING VALUES
df["Embarked"].isnull().sum()
```

```python
# find existing categories
df["Embarked"].value_counts()
```

```python
#we fill missing categorical values with mode
#finding mode
df["Embarked"].mode()[0]
```

```python
#filling the missing values
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
#checking for missing values in embarked
df["Embarked"].isnull().sum()
```

```python
#Understanding Cabin
df["Cabin"].isnull().sum()
```

```python
#creating a new category
df["Cabin"] = df["Cabin"].fillna("Unknown")
```

```python
#check for missing values
df["Cabin"].isnull().sum()
```

```python
#FINDING SUSPICIOUS VALUES
# min and max can help us to spot unusual values
df["Age"].describe
```

```python
#CHECCKING CATEGORICAL INCONSISTENCES
df["Sex"].value_counts()
```

```python
df["Sex"] =df["Sex"].str.lower()
```

```python
#REMOVING UNNECESSARY COLUMNS
#let us remove passenger id
df =df.drop(columns = ["PassengerId"])
```
